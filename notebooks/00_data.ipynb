# Data
Two separate paper-aligned tracks. Each has 20 first-order pilot questions and two separate debugging stories. Select `TRACK` below.
ThoughtTracing uses ParaphrasedToMi free-text answers. AutoToM uses its released ToMi container choices.

```python
import os
import pandas as pd
from tom_research.study import load_study, prepare_pilot, track_paths

TRACK = os.environ.get('TOM_TRACK', 'thoughttracing')
source, manifest_path = track_paths(TRACK)
prepare_pilot(source, manifest_path, track=TRACK)
debug, pilot, manifest = load_study(source, manifest_path, track=TRACK)
print(TRACK, manifest['dataset'], manifest['answer_format'])
pd.DataFrame([{'split': split, 'id': item.id, 'belief': item.belief_type, 'question': item.question}
              for split, items in [('debug', debug), ('pilot', pilot)] for item in items])
```

```python
pd.Series([item.belief_type for item in pilot]).value_counts().rename('questions')
```

```python
example = debug[0]
print(example.context)
print('Question:', example.question)
print('Debug answer:', example.answer)
print('Source hash:', manifest['source_sha256'])
```

## Answer Format
In the AutoToM track, both runs receive the dataset's container choices. In the ThoughtTracing track, neither run receives answer choices.
The free-text scorer checks for a gold-answer substring; the choice scorer requires one exact container answer. These pilot subsets are not full-paper evaluations.

```python
if TRACK == 'autotom':
    display(pd.DataFrame([{'id': item.id, 'supplied_choices': item.choices} for item in debug]))
else:
    print('Free-text answers; candidates are not shown to either model.')
```
